```python
geolocator = Nominatim(user_agent="shahzaib881")

def get_lat_lon(city):
    try:
        location = geolocator.geocode(city)
        if location:
            try:
                lat = location.latitude
            except:
                lat = None
            try:
                lon = location.longitude
            except:
                lon = None
            try:
                ad_len = len(location.address.split(", "))
            except:
                ad_len = 0
            try:
                cn = location.address.split(", ")[0]
            except:
                cn = None
            try:
                if ad_len > 5:
                    ct = location.address.split(", ")[-4]
                else:
                    ct = location.address.split(", ")[1]
            except:
                ct = None
            try:
                if ad_len > 5:
                    st = location.address.split(", ")[-3]
                elif ad_len < 5:
                    st = None
                else:
                    st = location.address.split(", ")[2]
            except:
                st = None
            try:
                if ad_len > 5:
                    zp = location.address.split(", ")[-2]
                elif ad_len < 5:
                    zp = None
                else:
                    zp = location.address.split(", ")[3]
                try:
                    zp = int("".join(re.findall('[0-9]+', str(zp))))
                except:
                    ct = st
                    st = zp
                    zp = None
            except:
                zp = None
            return ([lat, lon], cn, ct, st, zp)
    except Exception as e:
        print(f"Error getting coordinates for {city}: {e}")
    return None
```

```python
locss = l_counts['Location'].apply(get_lat_lon)
```

```python
l_counts['coordinates'] = locss.apply(lambda x: x[0] if x != None else None)
l_counts['correct_name'] = locss.apply(lambda x: x[1] if x != None else None)
l_counts['city_name'] = locss.apply(lambda x: x[2] if x != None else None)
l_counts['state_name'] = locss.apply(lambda x: x[3] if x != None else None)
l_counts['zip_code'] = locss.apply(lambda x: x[4] if x != None else None)
```

```python
world_map = folium.Map(location=[51.1657, 10.4515], zoom_start=6)
heat_data = [[coords[0], coords[1], count] for coords, count in zip(nlcount['coordinates'], nlcount['count'])]
HeatMap(heat_data).add_to(world_map)

# Add city markers with count as popups or tooltips
for i, row in nlcount.iterrows():
    lat, lon = row['coordinates']
    count = row['count']
    city = row['Location']
    folium.CircleMarker(
        location=[lat, lon],
        radius=3,  # Adjust marker size (in pixels)
        color='blue',
        fill=True,
        fill_color='blue',
        fill_opacity=0.6,
        popup=f"{city}: {count}",  # Shows when clicked
        tooltip=f"{city}: {count}"  # Shows when hovered
    ).add_to(world_map)

world_map.save("city_heatmap.html")
```

```python
import branca.colormap as cm
```

```python
# Create a base folium map (centered on Germany, adjust location and zoom if needed)
m = folium.Map(location=[51.1657, 10.4515], zoom_start=6)

# Create a color map based on the 'count' column
min_count = check_notna['count'].min()
max_count = check_notna['count'].max()
colormap = cm.LinearColormap(colors=['blue', 'green', 'yellow', 'red'], vmin=min_count, vmax=max_count)

# Add the city boundaries to the map using GeoJson and dynamically color them based on 'count'
folium.GeoJson(
    check_notna,
    style_function=lambda feature: {
        'fillColor': colormap(feature['properties']['count']),  # Dynamic color based on count
        'color': 'black',  # Boundary color
        'weight': 2,       # Boundary thickness
        'fillOpacity': 0.6,  # Opacity of the fill
    },
    tooltip=folium.GeoJsonTooltip(fields=['city_name', 'count'])  # Tooltip showing city name and count
).add_to(m)

# Add the color map legend
colormap.caption = 'City Count'
colormap.add_to(m)

# Save the map to an HTML file
m.save("city_boundaries_colored.html")
```
